```python
from sklearn.datasets import make_classification
import numpy as np

X, y = make_classification(n_samples=170, n_features=3, n_informative=1,n_redundant=0,
                           n_classes=2, n_clusters_per_class=1, random_state=41,hypercube=False,class_sep=15)
```

```python
import matplotlib.pyplot as plt

fig = plt.figure(figsize=(10, 6))
ax = fig.add_subplot(111, projection='3d')

ax.scatter(
    X[:, 0],
    X[:, 1],
    X[:, 2],
    c=y,
    cmap='winter',
    s=100
)

# Axis labels
ax.set_xlabel('Feature 1')
ax.set_ylabel('Feature 2')
ax.set_zlabel('Feature 3')

plt.show()
```

```python
def perceptron(X,y):

    w1=w2=w3=b=1
    learning_rate = 0.1

    for j in range(1000):

        for i in range(X.shape[0]):

            # check condition
            z = w1*X[i][0] + w2*X[i][1] + w3*X[i][2] + b

            if z*y[i] < 0:
                w1 = w1 + learning_rate*y[i]*X[i][0]
                w2 = w2 + learning_rate*y[i]*X[i][1]
                w3 = w3 + learning_rate*y[i]*X[i][2]
                b = b + learning_rate*y[i]

    return w1,w2,w3,b

```

```python
w1,w2,w3,b = perceptron(X,y)
print(w1)
print(w2)
print(w3)
print(b)
```

```python
x_input = np.linspace(X[:, 0].min(), X[:, 0].max(), 50)
y_input = np.linspace(X[:, 1].min(), X[:, 1].max(), 50)

# Create meshgrid
xx, yy = np.meshgrid(x_input, y_input)

# Calculate z values of decision plane
zz = -(w1 * xx + w2 * yy + b) / w3

# Create 3D figure
fig = plt.figure(figsize=(10, 7))
ax = fig.add_subplot(111, projection='3d')

# Plot decision boundary plane
ax.plot_surface(
    xx,
    yy,
    zz,
    alpha=0.4
)

# Plot dataset
ax.scatter(
    X[:, 0],
    X[:, 1],
    X[:, 2],
    c=y,
    cmap='winter',
    s=100
)

# Labels
ax.set_xlabel('Feature 1')
ax.set_ylabel('Feature 2')
ax.set_zlabel('Feature 3')

plt.show()
```
