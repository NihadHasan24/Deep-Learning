```python
from sklearn.datasets import make_classification
import numpy as np

X, y = make_classification(n_samples=500, n_features=2, n_informative=1,n_redundant=0,
                           n_classes=2, n_clusters_per_class=1, random_state=41,hypercube=False,class_sep=10)
```

```python
import matplotlib.pyplot as plt

plt.figure(figsize=(12,8))
plt.scatter(X[:,0],X[:,1],c=y,cmap='winter',s=100)
```

```python
def step(z):
    return 1 if z>0 else 0
```

```python
def perceptron(X,y):

    X = np.insert(X,0,1,axis=1)
    weights = np.ones(X.shape[1])
    learning_rate = 0.1

    for i in range(1000):
        random_point = np.random.randint(0,500)
        y_pred = step(np.dot(X[random_point],weights))
        weights = weights + learning_rate*(y[random_point]-y_pred)*X[random_point]

    return weights[0],weights[1:]
```

```python
intercept_,coef_ = perceptron(X,y)
print(coef_)
print(intercept_)
```

```python
m = -(coef_[0]/coef_[1])
b = -(intercept_/coef_[1])

x_input = np.linspace(-3,3,100)
y_input = m*x_input + b

plt.figure(figsize=(10,6))
plt.plot(x_input,y_input,color='red',linewidth=3)
plt.scatter(X[:,0],X[:,1],c=y,cmap='winter',s=100)
plt.ylim(-3,2)
```
