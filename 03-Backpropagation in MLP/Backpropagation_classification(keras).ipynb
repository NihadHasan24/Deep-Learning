```python
import numpy as np
import pandas as pd
```

```python
df = pd.DataFrame([[8,8,1],[7,9,1],[6,10,0],[5,12,0]],columns = ['CGPA','Profile_score','placed'])
```

```python
df
```

```python
import tensorflow
from tensorflow import keras
from tensorflow.keras import Sequential
from tensorflow.keras.layers import Dense
```

```python
from pandas.core.tools.datetimes import ArrayConvertible
model  = Sequential()

model.add(Dense(2,activation='sigmoid',input_dim=2))
model.add(Dense(1,activation='sigmoid'))
```

```python
model.summary()
```

```python
model.get_weights()
```

```python
new_wights = [np.array([[ 0.1,0.1],
        [0.1 , 0.1 ]], dtype=np.float32),
     np.array([0., 0.], dtype=np.float32),
     np.array([[0.1],
        [ 0.1]], dtype=np.float32),
     np.array([0.], dtype=np.float32)]
```

```python
model.set_weights(new_wights)
model.get_weights()
```

```python
optimizer = keras.optimizers.Adam(learning_rate=0.001)
model.compile(loss='binary_crossentropy',optimizer= optimizer)
```

```python
model.fit(df.iloc[:,0:-1].values,df['placed'].values,epochs=75,verbose=1,batch_size =1)
```
